#### imports et installation

```python
import cv2
import numpy as np
import os
```

#### chargement du detecteur MTCNN

```python
from mtcnn import MTCNN
detector = MTCNN()
```

#### Fonction de capture

```python
def add_personne(personne_nom, nb_images=100):
    dataset_path  = "dataset"
    personne_path = os.path.join(dataset_path, personne_nom)
    os.makedirs(personne_path, exist_ok=True)

    cap   = cv2.VideoCapture("http://192.168.10.122:8080/video")
    count = 0

    while True:
        ret, frame = cap.read()

        if not ret:
            print("Erreur de capture")
            break

        frame = cv2.resize(frame, (640, 480))

        # MTCNN travaille en RGB
        rgb    = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        faces  = detector.detect_faces(rgb)

        cv2.putText(frame, f"Images: {count}/{nb_images}", (10, 30),
                    cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)

        for face in faces:
            x, y, w, h = face['box']
            confidence  = face['confidence']

            # On ignore les détections peu sûres
            if confidence < 0.98:
                continue

            x, y = max(0, x), max(0, y)

            cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)

            face_img = frame[y:y+h, x:x+w]

            if face_img.size == 0:
                continue

            face_img = cv2.resize(face_img, (128, 128))
            img_name = os.path.join(personne_path, f"{count}.jpg")
            cv2.imwrite(img_name, face_img)
            count += 1
            print(f"Image {count} sauvegardée")
            break

        cv2.imshow("Capture", frame)

        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

        if count >= nb_images:
            break

    cap.release()
    cv2.destroyAllWindows()
    print(f"Terminé : {count} images dans '{personne_path}'")
```

### lancement

```python
add_personne(personne_nom="POLLASTRI")
```
